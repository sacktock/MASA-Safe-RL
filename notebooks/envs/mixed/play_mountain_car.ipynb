# Play MountainCar

Choose an environment with the buttons, preview an `rgb_array` frame inline, then run the last cell to play in a pygame window. Set `RECORD_VIDEO = True` to save the environment canvas as a 1920x1080 MP4.

```python
import ipywidgets as widgets
import numpy as np
from IPython.display import display
from PIL import Image
from masa.common.rendering.notebook_play import make_reset_env, sync_selected_env
from masa.common.rendering.notebook_play import start_recorded_play_thread

VALID_ENV_NAMES = ("DiscreteMountainCar", "ContinuousMountainCar")


def make_env(env_name, **kwargs):
    if env_name == "DiscreteMountainCar":
        from masa.envs.discrete.mountain_car import DiscreteMountainCar

        return DiscreteMountainCar(**kwargs)
    if env_name == "ContinuousMountainCar":
        from masa.envs.continuous.mountain_car import ContinuousMountainCar

        return ContinuousMountainCar(**kwargs)
    raise ValueError(f"env_name must be one of {VALID_ENV_NAMES!r}")


ENV_SELECTOR = widgets.ToggleButtons(
    options=VALID_ENV_NAMES, value="ContinuousMountainCar", description="Env"
)
display(ENV_SELECTOR)
ENV_NAME = ENV_SELECTOR.value
SEED = 0

RECORD_VIDEO = False
VIDEO_PATH = "videos/notebooks/play_mountain_car.mp4"
```

```python
ENV_NAME = ENV_SELECTOR.value
preview_env = make_env(ENV_NAME, render_mode="rgb_array", render_window_size=512)
obs, info = preview_env.reset(seed=SEED)
print("reset:", {"obs": obs, "info": info})
display(Image.fromarray(preview_env.render()))
preview_env.close()
```

```python
def action_from_pressed(env_name, pressed, pygame):
    if pressed[pygame.K_LEFT]:
        return (
            0
            if env_name == "DiscreteMountainCar"
            else np.array([-1.0], dtype=np.float32)
        )
    if pressed[pygame.K_RIGHT]:
        return (
            1
            if env_name == "DiscreteMountainCar"
            else np.array([1.0], dtype=np.float32)
        )
    return None


def _print_reset(obs, info):
    print("reset:", {"obs": obs, "info": info})


def play_env(env_name=None, seed=SEED):
    follow_selector = env_name is None

    def _run(stop_event):
        import pygame

        selected_env_name = env_name
        if follow_selector:
            selected_env_name = ENV_SELECTOR.value
        env, obs, info = make_reset_env(
            make_env,
            selected_env_name,
            seed=seed,
            render_mode="human",
            render_window_size=512,
        )
        finished = False
        print(
            "Controls: hold Left for negative force, hold Right for positive force, R resets, Q or Escape quits."
        )
        _print_reset(obs, info)
        try:
            running = True
            while running and not stop_event.is_set() and not env.human_window_closed:
                if follow_selector:
                    env, selected_env_name, obs, info, switched = sync_selected_env(
                        env,
                        selected_env_name,
                        ENV_SELECTOR,
                        make_env,
                        seed=seed,
                        render_window_size=512,
                        pygame=pygame,
                    )
                    if switched:
                        finished = False
                        print("switched:", selected_env_name)
                        _print_reset(obs, info)
                for event in pygame.event.get():
                    if not env.handle_pygame_event(event):
                        running = False
                        break
                    if event.type != pygame.KEYDOWN:
                        continue
                    if event.key in (pygame.K_q, pygame.K_ESCAPE):
                        running = False
                        break
                    if event.key == pygame.K_r:
                        obs, info = env.reset(seed=seed)
                        finished = False
                        _print_reset(obs, info)
                        continue
                if not running or stop_event.is_set():
                    break
                action = (
                    None
                    if finished
                    else action_from_pressed(
                        selected_env_name, pygame.key.get_pressed(), pygame
                    )
                )
                if action is None:
                    env.render()
                    continue
                obs, reward, terminated, truncated, info = env.step(action)
                finished = terminated or truncated
                print(
                    {
                        "obs": obs,
                        "reward": reward,
                        "terminated": terminated,
                        "truncated": truncated,
                        "info": info,
                    }
                )
        finally:
            env.close()

    return start_recorded_play_thread(
        "mountain_car", _run, record_video=RECORD_VIDEO, video_path=VIDEO_PATH
    )


play_session = play_env(seed=SEED)
```
